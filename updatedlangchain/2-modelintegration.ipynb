### Models Integration with OpenAI, Gemini, Groq

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python
from langchain.chat_models import init_chat_model

model = init_chat_model("gpt-4.1")
model
```

```python
# Invoke the model
response = model.invoke("Hello, how are you?")
response
```

```python
response.content
```

```python
# ChatOpenAI
from langchain_openai import ChatOpenAI

model = ChatOpenAI(model = "gpt-4.1")
response = model.invoke("Hello, how are you?")
response.content
```

### Google Gemini Model Integration

```python
from langchain.chat_models import init_chat_model

model = init_chat_model("google_genai:gemini-3.5-flash-lite")
response = model.invoke("Hello, how are you?")
response.content
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(model = "gemini-3.5-flash-lite")
response = model.invoke("Why do parrots talk?")
response.content
```

### Groq Model Integration

```python
from langchain.chat_models import init_chat_model

model = init_chat_model("groq:openai/gpt-oss-20b")
response = model.invoke("Why do parrots talk?")
response.content
```

```python
from langchain_groq import ChatGroq

model = ChatGroq(model = "openai/gpt-oss-20b")
response = model.invoke("Hello, how are you?")
response.content
```

### Streaming and Batching

```python
model.invoke("Write me a 200 word paragraph on AI").content
```

```python
model.stream("Write me a 200 word paragraph on AI")
```

```python
for chunk in model.stream("Write me a 200 word paragraph on AI"):
    if chunk.text:
        print(chunk.text, end="", flush=True)
```

```python
model.batch(
    ["Why do parrots talk?",
    "Write me a 200 word paragraph on AI",
    "What is quantum computing?"],
    config = {
        'max_concurrency': 5 # Limit to 5 parallel calls
    }
)
```
